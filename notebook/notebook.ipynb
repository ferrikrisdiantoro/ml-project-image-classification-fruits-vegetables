```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt

# Define device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Data augmentation and normalization
train_transforms = transforms.Compose([
    transforms.RandomResizedCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

val_transforms = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

# Load dataset from your drive
train_dataset = datasets.ImageFolder('/content/sample_data/fruits-vegetables/train', transform=train_transforms)
val_dataset = datasets.ImageFolder('/content/sample_data/fruits-vegetables/valid', transform=val_transforms)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
```

# **4. Declare and Training Base Models**

## **4.1. Swin Transformer**

```python
from transformers import SwinForImageClassification
from torch.optim import AdamW

from torch.optim.lr_scheduler import StepLR
import torch.nn as nn

# Load Swin Transformer model
swin_model = SwinForImageClassification.from_pretrained("microsoft/swin-base-patch4-window7-224")
swin_model.classifier = nn.Linear(swin_model.config.hidden_size, len(train_dataset.classes))  # Adjust classifier layer
swin_model.to(device)

# Define optimizer and scheduler
swin_optimizer = AdamW(swin_model.parameters(), lr=1e-4)
swin_scheduler = StepLR(swin_optimizer, step_size=5, gamma=0.1)

# Training function for Swin Transformer
def train_model(model, optimizer, scheduler, train_loader, val_loader, num_epochs=10):
    train_losses, train_accuracies, val_losses, val_accuracies = [], [], [], []

    for epoch in range(num_epochs):
        model.train()
        running_loss = 0.0
        correct_train = 0
        total_train = 0

        # Training phase
        for images, labels in train_loader:
            images, labels = images.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(images).logits
            loss = nn.CrossEntropyLoss()(outputs, labels)
            loss.backward()
            optimizer.step()

            running_loss += loss.item() * images.size(0)
            _, preds = torch.max(outputs, 1)
            correct_train += (preds == labels).sum().item()
            total_train += labels.size(0)

        # Learning rate scheduling
        scheduler.step()

        # Validation phase
        model.eval()
        val_loss, correct_val = 0.0, 0
        total_val = 0
        with torch.no_grad():
            for images, labels in val_loader:
                images, labels = images.to(device), labels.to(device)
                outputs = model(images).logits
                loss = nn.CrossEntropyLoss()(outputs, labels)
                val_loss += loss.item() * images.size(0)
                _, preds = torch.max(outputs, 1)
                correct_val += (preds == labels).sum().item()
                total_val += labels.size(0)

        # Calculate metrics per epoch
        train_loss = running_loss / len(train_loader.dataset)
        val_loss = val_loss / len(val_loader.dataset)
        train_accuracy = correct_train / total_train
        val_accuracy = correct_val / total_val

        train_losses.append(train_loss)
        train_accuracies.append(train_accuracy)
        val_losses.append(val_loss)
        val_accuracies.append(val_accuracy)

        # Print epoch results
        print(f"Epoch {epoch+1}/{num_epochs}, "
              f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_accuracy:.4f}, "
              f"Val Loss: {val_loss:.4f}, Val Accuracy: {val_accuracy:.4f}")

    return train_losses, train_accuracies, val_losses, val_accuracies

# Train the Swin Transformer
swin_train_losses, swin_train_accuracies, swin_val_losses, swin_val_accuracies = train_model(
    swin_model, swin_optimizer, swin_scheduler, train_loader, val_loader, num_epochs=10)
```

```python
import matplotlib.pyplot as plt

def plot_metrics(train_losses, val_losses, train_accuracies, val_accuracies, title="Model Training Progress"):
    epochs = range(1, len(train_losses) + 1)
    plt.figure(figsize=(14, 6))

    # Losses
    plt.subplot(1, 2, 1)
    plt.plot(epochs, train_losses, 'bo-', label='Training Loss')
    plt.plot(epochs, val_losses, 'ro-', label='Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.title(f'{title} - Loss')
    plt.legend()

    # Accuracy
    plt.subplot(1, 2, 2)
    plt.plot(epochs, train_accuracies, 'go-', label='Training Accuracy')
    plt.plot(epochs, val_accuracies, 'ro-', label='Validation Accuracy')  # Changed to 'ro-' for validation
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.title(f'{title} - Accuracy')
    plt.legend()

    plt.show()

# Call the plot function for Swin Transformer with correct arguments
plot_metrics(swin_train_losses, swin_val_losses, swin_train_accuracies, swin_val_accuracies, title="Swin Transformer")
```

```python
import torchvision.models as models
import torch.nn as nn
from torch.optim import AdamW

from torch.optim.lr_scheduler import StepLR

# Load ResNet model
resnet_model = models.resnet50(pretrained=True)
resnet_model.fc = nn.Linear(resnet_model.fc.in_features, len(train_dataset.classes))  # Adjust classifier layer
resnet_model.to(device)

# Define optimizer and scheduler for ResNet
resnet_optimizer = AdamW(resnet_model.parameters(), lr=1e-4)
resnet_scheduler = StepLR(resnet_optimizer, step_size=5, gamma=0.1)

# Modify the train_model function to handle models that output raw tensors instead of a logits attribute
def train_resnet_model(model, optimizer, scheduler, train_loader, val_loader, num_epochs=10):
    train_losses, train_accuracies, val_losses, val_accuracies = [], [], [], []

    for epoch in range(num_epochs):
        model.train()
        running_loss = 0.0
        correct_train = 0
        total_train = 0

        # Training phase
        for images, labels in train_loader:
            images, labels = images.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(images)  # Direct output without `.logits` for ResNet
            loss = nn.CrossEntropyLoss()(outputs, labels)
            loss.backward()
            optimizer.step()

            running_loss += loss.item() * images.size(0)
            _, preds = torch.max(outputs, 1)
            correct_train += (preds == labels).sum().item()
            total_train += labels.size(0)

        # Learning rate scheduling
        scheduler.step()

        # Validation phase
        model.eval()
        val_loss, correct_val = 0.0, 0
        total_val = 0
        with torch.no_grad():
            for images, labels in val_loader:
                images, labels = images.to(device), labels.to(device)
                outputs = model(images)  # Direct output without `.logits` for ResNet
                loss = nn.CrossEntropyLoss()(outputs, labels)
                val_loss += loss.item() * images.size(0)
                _, preds = torch.max(outputs, 1)
                correct_val += (preds == labels).sum().item()
                total_val += labels.size(0)

        # Calculate epoch metrics
        train_loss = running_loss / len(train_loader.dataset)
        train_accuracy = correct_train / total_train
        val_loss = val_loss / len(val_loader.dataset)
        val_accuracy = correct_val / total_val

        train_losses.append(train_loss)
        train_accuracies.append(train_accuracy)
        val_losses.append(val_loss)
        val_accuracies.append(val_accuracy)

        # Print epoch results
        print(f"Epoch {epoch+1}/{num_epochs}, "
              f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_accuracy:.4f}, "
              f"Val Loss: {val_loss:.4f}, Val Accuracy: {val_accuracy:.4f}")

    return train_losses, train_accuracies, val_losses, val_accuracies

# Train the ResNet model using the modified train function
resnet_train_losses, resnet_train_accuracies, resnet_val_losses, resnet_val_accuracies = train_resnet_model(
    resnet_model, resnet_optimizer, resnet_scheduler, train_loader, val_loader, num_epochs=10)

# Plot ResNet training progress
plot_metrics(resnet_train_losses, resnet_val_losses, resnet_train_accuracies, resnet_val_accuracies, "ResNet")
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Collect true labels from validation set
val_labels = []
with torch.no_grad():
    for _, labels in val_loader:
        val_labels.extend(labels.numpy())
val_labels = np.array(val_labels)

# Function to plot confusion matrix
def plot_confusion_matrix(labels, predictions, title="Confusion Matrix"):
    conf_matrix = confusion_matrix(labels, predictions)
    plt.figure(figsize=(8, 6))
    sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues",
                xticklabels=train_dataset.classes, yticklabels=train_dataset.classes)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title(title)
    plt.show()

# Evaluate Swin Transformer model individually
swin_val_predictions = []
with torch.no_grad():
    for images, _ in val_loader:
        images = images.to(device)
        outputs = swin_model(images).logits
        _, preds = torch.max(outputs, 1)
        swin_val_predictions.extend(preds.cpu().numpy())

print("Swin Transformer Classification Report:")
print(classification_report(val_labels, swin_val_predictions, target_names=train_dataset.classes))
plot_confusion_matrix(val_labels, swin_val_predictions, title="Swin Transformer Confusion Matrix")

# Evaluate ResNet model individually
resnet_val_predictions = []
with torch.no_grad():
    for images, _ in val_loader:
        images = images.to(device)
        outputs = resnet_model(images)
        _, preds = torch.max(outputs, 1)
        resnet_val_predictions.extend(preds.cpu().numpy())

print("ResNet Classification Report:")
print(classification_report(val_labels, resnet_val_predictions, target_names=train_dataset.classes))
plot_confusion_matrix(val_labels, resnet_val_predictions, title="ResNet Confusion Matrix")
```

```python
from sklearn.svm import SVC
import numpy as np
import torch

def extract_features(model, loader, device):
    model.eval()
    features, labels = [], []
    with torch.no_grad():
        for images, label in loader:
            images = images.to(device)

            # Ekstraksi fitur dari model Swin atau ResNet
            outputs = model(images)

            # Periksa apakah output berupa dictionary (misalnya untuk Swin)
            if isinstance(outputs, dict):
                feat = outputs['logits']  # Ambil tensor logits dari dictionary
            elif hasattr(outputs, 'logits'):
                feat = outputs.logits  # Untuk output model khusus (misalnya Swin)
            else:
                feat = outputs  # Untuk model biasa seperti ResNet yang langsung mengeluarkan tensor

            # Flatten output ke 1D vector dan pindahkan ke CPU
            feat = feat.view(feat.size(0), -1).cpu().numpy()
            features.append(feat)
            labels.append(label.cpu().numpy())

    return np.vstack(features), np.hstack(labels)


# Pindahkan model Swin dan ResNet ke GPU
swin_model.to(device)
resnet_model.to(device)

# Ekstrak fitur dari kedua model untuk data pelatihan
swin_train_features, train_labels = extract_features(swin_model, train_loader, device)
print(f"Swin Train Features Shape: {swin_train_features.shape}")

resnet_train_features, _ = extract_features(resnet_model, train_loader, device)
print(f"ResNet Train Features Shape: {resnet_train_features.shape}")

# Gabungkan fitur Swin dan ResNet
train_features = np.hstack((swin_train_features, resnet_train_features))
print(f"Combined Train Features Shape: {train_features.shape}")

# Ekstrak fitur untuk data validasi
swin_val_features, val_labels = extract_features(swin_model, val_loader, device)
print(f"Swin Val Features Shape: {swin_val_features.shape}")

resnet_val_features, _ = extract_features(resnet_model, val_loader, device)
print(f"ResNet Val Features Shape: {resnet_val_features.shape}")

# Gabungkan fitur validasi
val_features = np.hstack((swin_val_features, resnet_val_features))
print(f"Combined Val Features Shape: {val_features.shape}")

# Pastikan dimensi fitur yang digabungkan
print(f"Training Features Shape: {train_features.shape}")
print(f"Validation Features Shape: {val_features.shape}")

# Melatih meta-learner SVM dengan fitur gabungan
meta_learner = SVC(kernel='linear')
meta_learner.fit(train_features, train_labels)

# Evaluasi meta-learner pada data validasi
val_predictions = meta_learner.predict(val_features)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

# Print classification report and plot confusion matrix for meta-learner
print("Meta-Learner (SVM) Classification Report:")
print(classification_report(val_labels, val_predictions, target_names=train_dataset.classes))
plot_confusion_matrix(val_labels, val_predictions, title="Meta-Learner (SVM) Confusion Matrix")
```
